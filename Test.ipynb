```python
import pandas as pd
import re
import unicodedata
from data_loader import load_test_data

test_source1, test_source2, test_source3 = load_test_data(
    "../../../dataset"
)
```

```python
test_source1 = pd.read_csv(
    "/Users/aaravraj/Downloads/student_resource/dataset/test/test_source1.tsv",
    sep="\t"
)

test_source2 = pd.read_csv(
    "/Users/aaravraj/Downloads/student_resource/dataset/test/test_source2.tsv",
    sep="\t"
)

test_source3 = pd.read_csv(
    "/Users/aaravraj/Downloads/student_resource/dataset/test/test_source3.tsv",
    sep="\t"
)
```

```python
def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()

    text = unicodedata.normalize("NFKD", text)

    text = "".join(
        c for c in text
        if not unicodedata.combining(c)
    )

    text = text.replace("&", " and ")

    text = re.sub(r"[^a-z0-9\s]", " ", text)

    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
def clean_business_name(text):
    return normalize_text(text)
```

```python
def clean_address(text):

    if pd.isna(text):
        return ""

    text = str(text).lower().strip()

    text = unicodedata.normalize("NFKD", text)

    text = "".join(
        c for c in text
        if not unicodedata.combining(c)
    )

    text = text.replace("&", " and ")

    replacements = {
        r"\brd\.?\b": "road",
        r"\bst\.?\b": "street",
        r"\bave\.?\b": "avenue",
        r"\bav\.?\b": "avenue",
        r"\bblvd\.?\b": "boulevard",
        r"\bln\.?\b": "lane",
        r"\bdr\.?\b": "drive",
        r"\bhwy\.?\b": "highway"
    }

    for pattern, replacement in replacements.items():
        text = re.sub(pattern, replacement, text)

    text = re.sub(r"[^a-z0-9\s]", " ", text)

    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
def clean_country(text):

    if pd.isna(text):
        return ""

    text = str(text).lower().strip()

    text = re.sub(r"\s+", " ", text)

    return text
```

```python
for df in [test_source1, test_source2, test_source3]:

    df["business_name_clean"] = (
        df["business_name"]
        .apply(clean_business_name)
    )

    df["business_address_clean"] = (
        df["business_address"]
        .apply(clean_address)
    )

    df["country_clean"] = (
        df["country"]
        .apply(clean_country)
    )
```
